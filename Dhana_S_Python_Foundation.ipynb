```python
# Celsius to Fahrenheit Conversion
# F = (C × 9/5) + 32
celsius = 37.0
fahrenheit = (celsius * 9/5) + 32
print(f"{celsius}°C is equivalent to {fahrenheit}°F")
```

```python
#Fahrenheit to Celsius Conversion
# C = (F - 32) × 5/9
fahrenheit = 98.6
celsius = (fahrenheit - 32) * 5/9
print(f"{fahrenheit}°F is equivalent to {celsius}°C")
```

```python
#Swapping Two Variables (Temporary Variable)
x=5
y=10
print("Original: x =", x, "y =", y)
Temp = x
x=y
y=Temp  
print("After: x =", x, "y =", y)
```

```python
# Swapping Two Variables (Pythonic Way)
#Packing and Unpacking using a tuple
x=5
y=10
print("Original: x =", x, "y =", y)
Numbers=[x,y]
y,x=Numbers
print("After Swapping: x =", x, "y =", y)
```

```python
#Simple Interest Calculation
#simple interest SI = (P × R × T) / 100
principal = 1000
rate = 5
time = 2
SI = (principal * rate * time) / 100   
print("Simple Interest:", SI)
```

```python
#Compound Interest Calculation
#compound interest -- A = P × (1 + R/100)^T  and CI = A - P.
principal = 1200
rate = 5.4
time = 2
A = principal * (1 + rate / 100) ** time
CI = A - principal
print("Compound Interest: {:.2f}".format(CI))
```
