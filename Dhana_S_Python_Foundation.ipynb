```python
# Quadratic Equation Roots
"""
calculates the real or complex roots of a quadratic equation ax² + bx + c = 0 using the
determinant formula.
"""
a,b,c= 1,5,6
#a,b,c = 1,2,5
if a != 0:
    discriminant = b**2 - 4*a*c   
    real_part = -b / (2*a)     
    # Two distinct real roots
    if discriminant > 0:
        sqrt_discriminant = discriminant **.5
        root1 = (-b + sqrt_discriminant) / (2*a)
        root2 = (-b - sqrt_discriminant) / (2*a)   
        print(f"Two Distinct Real Roots: {root1} and {root2}")          
    #One repeated real root
    elif discriminant == 0:
        x = -b / (2*a)
        print(f"Repeated Root: {x}")               
    #Two complex conjugate roots
    else:
        imag_part = (-discriminant) ** 0.5 / (2*a)
        print(f"Complex Roots: {real_part} + {imag_part}i and {real_part} - {imag_part}i")
else:
    print("Not a quadratic equation")

            
```

```python
#Palindrome Check for Numbers
# Palindrome if a numeric value remains identical when its digits are reversed.
def ReverseNum(n):
    reversed_num = 0
    while n > 0:
        last_digit = n % 10
        reversed_num = reversed_num * 10 + last_digit
        n //= 10
    return reversed_num

def IsPalindrome(n):
    return n == ReverseNum(n)
Num=1221
print(f"IsPalindrome  : {IsPalindrome(Num)}")
```

```python
#Prime Number Check
#Number is prime : Number >1 and No divisors other than 1 and itself.
def isPrime(number):  
   prime=True
   if number<2:
      return False
   for i in range(2,number):  
      if number % i ==0:  
         prime=False  
         break  
   return prime 

Num = 29
print(f"Is {Num} prime?: {isPrime(Num)}")
```

```python
#Count Digits in an Integer
#Write a program that counts the total number of digits in a given integer.
def countDigits(n):
    count = 0
    while n > 0:
        n //= 10
        count += 1
    return count

Num = 98456
print(f"Number of digits in {Num}: {countDigits(Num)}") 
```

```python
#Find G.C.D (Greatest Common Divisor)
"""
Find the quotient Q and remainder R when A is divided by B, R= A % B
Replace A=B  B= R
 Repeat until = Do steps 2 and 3 until R == 0
 """
def gcd(a, b):
    while b != 0:
        a=b
        r= a % b
        b=r       
    return a

a, b =60,48
print(f"GCD : {gcd(a, b)}")
```

```python
#Armstrong Number Check
"""
An Armstrong number is a number that equals the sum of its own digits,
where each digit is raised to the power of the total number of digits.
1634 = 1^4 + 6^4 + 3^4 + 4^4 = 1634 isAmstrong Number
1421 = 1^4 + 4^4 + 2^4 + 1^4 = 354 is NOT Amstrong Number
"""
def is_armstrong(num):
    num_str = str(num)
    p= len(num_str)    
    # Calculate the sum of digits raised to the power    
    sum=0
    while num > 0:
        last_digit = num % 10
        sum += last_digit ** p
        num //= 10
        
    # Check if the sum =  original number
    
    return sum == int(num_str)

N  = 1634
print(f" Is {N} Armstrong?: {is_armstrong(N)}")
```
